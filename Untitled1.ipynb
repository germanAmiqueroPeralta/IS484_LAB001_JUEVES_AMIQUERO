```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22, margen=1.0):
    """
    el agente recibe la temperatura actual del ambiente , uede realizar  acciones de3 enfriar,
    calentar o mantener la temperatura. Su entorno es el ambiente donde se controla la temperatura. Su objetivo es
    mantener la temperatura alrededor de 22 °C. Su medida de desempeño consiste en mantener la temperatura cercana  
    a 22 °C,
    """
    diferencia = temperatura_actual - temperatura_objetivo
    if diferencia > margen:
        return "enfriar"
    elif diferencia < -margen:
        return "calentar"
    else:
        return "no realizar ninguna accion "
```
