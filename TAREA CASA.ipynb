```python
PEAS
Medida de desempeño (P): Se considera que el agente funciona bien si mantiene una humedad adecuada en el suelo.
Entorno (E): El entorno está formado por el jardín, el sistema de riego y las condiciones del clima.
Acciones (A): regar el jardín o no regarlo.
Percepción (S): Humedad del suelo y predicción so llovera durante el día.

Objetivo: El objetivo es mantener el suelo con una humedad adecuada.
```

```python
REGLAS DE DESICION
si el suelo está seco y no lloverá, entonces: regar
si el suelo está seco y lloverá, entonces: no regar
si la humedad es adecuada y lloverá, entonces: no regar
si  la humedad es adecuada y no lloverá entonces: no regar
```

```python
PARA VALORES CONCRETOS
Humedad menor a 30: suelo seco
Humedad entre 30 y 70: humedad adecuada
Humedad mayor a 70: saturado
```

```python
def agente_riego(humedad_suelo, llovera_hoy):
    if humedad_suelo < 30 and not llovera_hoy:
        return "Regar"
    else:
        return "No regar"


print(agente_riego(20, False))
print(agente_riego(20, True))
print(agente_riego(45, False))
print(agente_riego(75, False))
print(agente_riego(60, True))
print(agente_riego(10, True))
```
